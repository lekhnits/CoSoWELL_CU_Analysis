# Metrics of new picture descriptions

New Cookie Theft descriptions, measured with the same functions as the corpus (`results/features/`): language
metrics one text at a time; creativity metrics together with the whole corpus, because rarity and originality
compare a text with everyone else's. Texts under 20 words get no creativity metrics (the analysis leaves them out).

```python
import sys
from pathlib import Path

ROOT = Path.cwd().resolve()
sys.path.insert(0, str(ROOT / "src"))

import pandas as pd
from IPython.display import display

from cosowell_language.config import DATA_DIR, RESULTS_DIR
from cosowell_language.metrics.new_picture_descriptions import (measure_new_picture_descriptions, new_texts_table,
                                                                read_new_texts)

pd.set_option("display.max_columns", 30)
```

## 1. The new texts

Either a CSV file with a `narrative` column (and optionally `id`), or texts typed into `TYPED_TEXTS`. The file is
used when it exists.

```python
NEW_TEXTS_FILE = DATA_DIR / "new_cookie_texts.csv"

TYPED_TEXTS = [
    "In the kitchen a boy is standing on a stool that is about to fall over. He is taking cookies out of the jar "
    "and handing one to his sister, who is reaching up. Their mother is drying a plate and does not notice that "
    "the sink is overflowing onto the floor. Outside the window you can see the garden.",
]

if NEW_TEXTS_FILE.exists():
    new_texts = read_new_texts(NEW_TEXTS_FILE)
else:
    new_texts = new_texts_table(TYPED_TEXTS)
new_texts
```

## 2. Metrics

`WITH_CREATIVITY = False` skips the creativity metrics (a few minutes, and the transformer models).

```python
WITH_CREATIVITY = True

new_metrics = measure_new_picture_descriptions(new_texts, with_creativity=WITH_CREATIVITY)
print(len(new_metrics), "texts x", new_metrics.shape[1] - 3, "metrics")
```

## 3. Saved table

One row per text: id, text, then every metric, with the column names of `results/features/`.

```python
OUTPUT_FILE = RESULTS_DIR / "new_picture_descriptions" / "cookie_metrics.csv"

OUTPUT_FILE.parent.mkdir(parents=True, exist_ok=True)
new_metrics.to_csv(OUTPUT_FILE, index=False)
print("saved:", OUTPUT_FILE.relative_to(ROOT))
new_metrics.drop(columns=["test_session", "narrative"]).set_index("id").T.round(3)
```
